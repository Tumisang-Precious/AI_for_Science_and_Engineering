Pretrained model name: TabPFN-v2-reg (Tabular Prior-Data Fitted Network – Regression)

link:https://huggingface.co/Prior-Labs/TabPFN-v2-reg?utm

TabPFN is a transformer-based tabular foundation model. It is designed to identify relationships between variables in structured datasets and make predictions, particularly on relatively small datasets.


- Input: Numerical soil properties, such as liquid limit, plasticity index, clay content, moisture content and other relevant geotechnical measurements.
- Processing: The pretrained transformer analyses patterns and relationships between the input variables and the target variable.
- Output: A predicted continuous value, such as the percentage of free swelling of a soil sample.
The model is designed for regression and can make predictions without requiring the conventional task-specific parameter training used by many machine learning algorithms.

**Application in geological and geotechnical engineering**

TabPFN could be useful in geotechnical engineering for predicting soil behaviour from laboratory test results.

Potential applications include:
1. Predicting soil swelling potential from index properties.
2. Supporting preliminary identification of expansive soils.
3. Assessing relationships between soil plasticity, clay content and swelling behaviour.
4. Supporting preliminary foundation investigations where problematic soils may cause differential movement and structural cracking.
In your research, the model could provide an additional method of predicting swelling percentage from laboratory soil measurements. Such predictions could eventually support the assessment of potential foundation-related problems.

**Difference between TabPFN and the machine learning models developed during the course**

The main difference is that TabPFN is a pretrained foundation model, whereas the conventional supervised learning models developed during the course learn their predictive relationships directly from the training dataset.

| Aspect | Conventional ML models | TabPFN-v2-reg |
| :--- | :--- | :--- |
| **Model type** | Algorithms such as Random Forest or SVR | Transformer-based foundation model |
| **Training** | Learns from the project's training data | Pretrained on synthetic tabular tasks |
| **Learning approach** | Task-specific model fitting | Uses pretrained knowledge and supplied training examples |
| **Data requirements** | Depends on the algorithm and dataset | Particularly suited to smaller tabular datasets |
| **Application** | Learns patterns specific to the supplied dataset | Uses learned prior patterns to support predictions on new tabular tasks |

An important distinction is that TabPFN is not simply a model that predicts without seeing data. It still requires labelled examples from the project dataset to establish the relationship between soil properties and swelling percentage. However, it uses knowledge acquired during pretraining rather than learning all its underlying parameters from scratch for each new project.

 **Limitations and precautions**

Before applying TabPFN to an engineering investigation,the following should be considered:
1. TabPFN is particularly suited to relatively small tabular datasets. Its suitability must be checked against the size and characteristics of the available soil dataset.
2. Incorrect laboratory measurements, missing values and inconsistent units could affect predictions.
3. A model developed using one collection of soil samples may not perform equally well on soils from different geological formations or geographical locations.
4. Predictions must be evaluated using independent test data and appropriate regression metrics, including MAE, RMSE and \(R^2\).
5. Predictions should be compared with laboratory measurements and established geotechnical testing methods.
6. The model should not independently determine foundation designs or replace professional geotechnical investigations.
7. The model requires suitable computing resources and compatible software. Its model card specifies Python 3.9 or later, PyTorch 2.1 or later, and recommends 16 GB or more of RAM.
