Project problem: Determine whether measurable clay-soil properties can be used to predict swelling behaviour that may indicate an increased risk of foundation movement and premature building cracking.

Intended AI/ML outcome: Develop a machine-learning regression model that predicts soil swelling potential / swell strain from routinely measured geotechnical properties. The predicted swelling behaviour will be treated as an intermediate geotechnical risk indicator for foundation movement and potential cracking.

Important limitation: The selected public dataset contains soil swelling measurements, not direct observations of building cracks.

For the complete research problem, the ideal dataset would combine three groups of information:

Clay/soil properties:	clay content, liquid limit (LL), plastic limit (PL), plasticity index (PI), activity, moisture content, void ratio, dry/unit weight

Foundation/building information:	foundation type, foundation depth, building age, construction quality, foundation dimensions, drainage condition

Observed factors:	settlement, foundation movement, crack width, crack length, crack pattern, crack severity, crack location

Dataset: Data on one-dimensional vertical free swelling potential of soils and related soil properties

Source: GitHub repository maintained by the dataset authors.

The dataset was compiled from geotechnical studies and contains 517 records of soil swelling. The published study reports soil inputs including moisture content, void ratio, liquid limit, plasticity index, unit weight, clay/fine/coarse content, maximum dry unit weight and activity. The data were collected from soils of different plasticity and swelling behaviour.

Why this dataset was selected: It is much closer to the project problem than a generic building-crack image dataset because it measures the geotechnical mechanism that can cause foundation movement.

Dataset repository:
https://github.com/onyekpeu/Data-on-one-dimensional-vertical-free-swelling-potential-of-soils-and-related-soil-properties-

Related research article:
https://doi.org/10.1016/j.gsf.2021.101296

````python
import pandas as pd
import numpy as np
import requests
from io import BytesIO

DATA_URL = (
    "https://github.com/onyekpeu/Data-on-one-dimensional-vertical-free-swelling-potential-of-soils-and-related-soil-properties-/blob/main/Dataset.xlsx?raw=true"
)

response = requests.get(DATA_URL)
response.raise_for_status()

excel_file = pd.ExcelFile(BytesIO(response.content))
print("Available sheets:", excel_file.sheet_names)

df = pd.read_excel(BytesIO(response.content), sheet_name=excel_file.sheet_names[0])

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head(10))
````

````python
# Basic inspection

display(df.head())
display(df.tail())

print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

print("\nData types:")
display(df.dtypes)

print("\nMissing values:")
display(df.isna().sum().sort_values(ascending=False))
````
